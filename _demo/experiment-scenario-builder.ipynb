```python
import itertools
```

```python
study_variants = ["full", "ablation"]
schemas = ["dwc", "eml"]
strategies = ["data-focus", "author-focus"]
configurations = [
    dict(
        provider="openai-platform",
        core_platform="openai", 
        proxy_platform="openai", 
        core_llm="gpt-5.4-mini", 
        embedding_model="text-embedding-3-small", 
        rag_llm="gpt-5.4-mini"
    ),
    dict(
        provider="surf",
        core_platform="surf-pilot", 
        proxy_platform="openai", 
        core_llm="gpt-oss-120b", 
        embedding_model="text-embedding-3-small", 
        rag_llm="gpt-5.4-mini"
    ),
    dict(
        provider="surf",
        core_platform="surf-pilot", 
        proxy_platform="openai", 
        core_llm="mistral-small-3.2", 
        embedding_model="text-embedding-3-small", 
        rag_llm="gpt-5.4-mini"
    ),
    dict(
        provider="ivi",
        core_platform="surf-ivi", 
        proxy_platform="openai", 
        core_llm="o3-mini", 
        embedding_model="text-embedding-3-small", 
        rag_llm="gpt-5.4-mini"
    ),
]
```

```python
scenarios = list(itertools.product(study_variants, schemas, strategies, configurations))
```

```python
import pandas as pd
```

```python
scenario_records = [
    {
        "variant": s[0],
        "schema": s[1],
        "strategy": s[2],
        "provider": s[3]["provider"],
        "config": s[3],
    } for s in scenarios
]
```

```python
scenarios_df = pd.DataFrame.from_records(scenario_records)
```

```python
scenarios_df.describe(include="all")
```

```python
scenarios_base_platform = scenarios_df[scenarios_df["config"].apply(lambda c: c["core_platform"] == "openai")]
scenarios_base_platform
```

```python
scenarios_base_platform_full = scenarios_base_platform[scenarios_base_platform["variant"]=="full"]
scenarios_base_platform_full
```
