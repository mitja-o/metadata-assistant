```python
from abc import ABC, abstractmethod

class PromptClient(ABC):
    @abstractmethod
    def prompt_user(
        self, queries: list[str] = [], prelude: list[str] = []
    ) -> list[str]: ...



class StandardIOPromptClient(PromptClient):
    def prompt_user(self, queries=[], prelude=[]):
        for context_msg in prelude:
            print(context_msg)
        return [input(query) for query in queries]
```

```python
s_client = StandardIOPromptClient()
```

```python
s_client.prompt_user(["What's up???", "Tell Me More!"], ["Now that we're talking..."])
```

```python
import time
from ipywidgets import interact_manual, widgets

class JupyterTextWidgetIOPromptClient(PromptClient):
    def prompt_user(self, queries=[], prelude=[]):
        for msg in prelude: 
            print(msg)
        container = []
        for i, q in enumerate(queries):
            ui = interact_manual(
                lambda **kwargs: container.extend(kwargs.values()), 
                **{f"q_{i}": widgets.Textarea(description=q, style={"description_width": "initial"})}
            )

            while len(container) < i+1:
                time.sleep(0.1)

            for child in ui.widget.children:
                child.disabled = True

        return container
```

```python
j_client = JupyterTextWidgetIOPromptClient()
```

```python
j_client.prompt_user(["What's up???", "Tell Me More!"], ["Now that we're talking..."])
```
